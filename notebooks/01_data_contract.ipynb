# 01 — Safe restaurant data contract
Define an analysis boundary that works with synthetic records or a user-downloaded open dataset—never embedded credentials.

```python
from dataclasses import dataclass

@dataclass(frozen=True)
class ReviewRecord:
    business_id: str
    stars: float
    text: str

sample = ReviewRecord('demo-cafe', 4.0, 'Friendly staff and a bright dining room.')
assert 1 <= sample.stars <= 5 and sample.text.strip()
sample
```

## Production rule
Keep acquisition outside the analytics package. Validate schema, license and retention policy at the boundary; inject records into downstream functions.
