# 02 — Review-text analysis
A compact version of the historical text-processing experiments, using synthetic reviews.

```python
from collections import Counter
from restaurant_insights.text import normalize_text

reviews = [
    'Fresh pasta and friendly service',
    'Slow service but excellent dessert',
    'Bright room, fresh ingredients',
]
tokens = [token for review in reviews for token in normalize_text(review).split()]
Counter(tokens).most_common(8)
```
