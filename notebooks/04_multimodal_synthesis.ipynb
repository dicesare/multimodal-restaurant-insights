# 04 — Multimodal synthesis
Combine review and image evidence at restaurant level while keeping each feature explainable.

```python
restaurants = [
    {'business_id':'demo-cafe','mean_stars':4.3,'review_count':38,'mean_brightness':0.68,'image_count':12},
    {'business_id':'demo-bistro','mean_stars':3.6,'review_count':21,'mean_brightness':0.44,'image_count':7},
]
for row in restaurants:
    row['evidence_coverage'] = min(1.0, row['review_count']/30) * min(1.0, row['image_count']/10)
restaurants
```

## Interpretation
Coverage is shown beside insights so weak evidence is visible. A production dashboard should preserve drill-down, provenance and uncertainty rather than output a single opaque score.
